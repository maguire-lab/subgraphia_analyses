```python
import pandas as pd

# summarize metadata. When ARO and tax_ID_names are the same, keep the median path length row, discard the rest. If they are different, keep all rows. 
# Do nto discard NA rows
def summarize_metadata(df):
    summarized = df.groupby(['ARO', 'tax_ID_names'], dropna=False).agg({
        'path_length': 'median',
        'dataset': 'first',  # Keep the dataset name for each group
        'path_id': 'first'  # Keep the path_id for each group
    }).reset_index()
    return summarized

ww_metadata = pd.read_csv('./metadata.tsv', sep='\t')
# Add 'dataset' column 
ww_metadata['dataset'] = 'ERR14173773'

# Apply the summarization function to the metadata
summarized_metadata = summarize_metadata(ww_metadata)

# Save the summarized metadata to a new TSV file
summarized_metadata.to_csv('./summarized_metadata.tsv', sep='\t', index=False)

```
